## SEL 1 — Fungsi bisnis TokoPolines

```python
# SEL 1 - Fungsi bisnis TokoPolines
def validasi_jumlah(jumlah, stok):
    return (type(jumlah) is int and type(stok) is int
            and stok >= 0 and 1 <= jumlah <= 10
            and jumlah <= stok)

def hitung_subtotal(harga, jumlah):
    if (type(harga) is not int or type(jumlah) is not int
            or harga < 0 or jumlah < 0):
        raise ValueError("Harga dan jumlah harus integer nonnegatif")
    return harga * jumlah

def hitung_diskon(subtotal, login, voucher):
    if type(subtotal) is not int or subtotal < 0:
        raise ValueError("Subtotal harus integer nonnegatif")
    if login is True and subtotal >= 100000 and voucher == "POLINES10":
        return subtotal // 10
    return 0

def validasi_checkout(keranjang):
    return isinstance(keranjang, list) and len(keranjang) > 0

def validasi_jumlah_bug(jumlah, stok):
    return (type(jumlah) is int and type(stok) is int
            and stok >= 0 and 1 <= jumlah <= 11
            and jumlah <= stok)
```

## SEL 1B — Percobaan manual
Output acuan berurutan: True, False, 250000, 25000, False.

```python
print(validasi_jumlah(10, 15))
print(validasi_jumlah(11, 15))
print(hitung_subtotal(50000, 5))
print(hitung_diskon(250000, True, "POLINES10"))
print(validasi_checkout([]))
```

## SEL 2 — Suite utama: 28 unit test (TestJumlah, TestSubtotal, TestDiskon, TestCheckout)

```python
import unittest

class TestJumlah(unittest.TestCase):
    def test_01_minimum(self): self.assertTrue(validasi_jumlah(1, 15))
    def test_02_normal(self): self.assertTrue(validasi_jumlah(5, 15))
    def test_03_maksimum(self): self.assertTrue(validasi_jumlah(10, 15))
    def test_04_nol(self): self.assertFalse(validasi_jumlah(0, 15))
    def test_05_sebelas(self): self.assertFalse(validasi_jumlah(11, 15))
    def test_06_stok(self): self.assertFalse(validasi_jumlah(8, 5))
    def test_07_desimal(self): self.assertFalse(validasi_jumlah(2.5, 15))
    def test_08_negatif(self): self.assertFalse(validasi_jumlah(-1, 15))
    def test_09_boolean(self): self.assertFalse(validasi_jumlah(True, 15))
    def test_10_stok_negatif(self): self.assertFalse(validasi_jumlah(1, -1))

class TestSubtotal(unittest.TestCase):
    def test_11_satu(self): self.assertEqual(hitung_subtotal(50000, 1), 50000)
    def test_12_lima(self): self.assertEqual(hitung_subtotal(50000, 5), 250000)
    def test_13_sepuluh(self): self.assertEqual(hitung_subtotal(50000, 10), 500000)
    def test_14_nol(self): self.assertEqual(hitung_subtotal(50000, 0), 0)
    def test_15_harga_negatif(self):
        with self.assertRaises(ValueError): hitung_subtotal(-50000, 5)
    def test_16_jumlah_negatif(self):
        with self.assertRaises(ValueError): hitung_subtotal(50000, -1)

class TestDiskon(unittest.TestCase):
    def test_17_valid(self): self.assertEqual(hitung_diskon(250000, True, "POLINES10"), 25000)
    def test_18_batas(self): self.assertEqual(hitung_diskon(100000, True, "POLINES10"), 10000)
    def test_19_belum_login(self): self.assertEqual(hitung_diskon(250000, False, "POLINES10"), 0)
    def test_20_subtotal_kurang(self): self.assertEqual(hitung_diskon(99999, True, "POLINES10"), 0)
    def test_21_voucher_salah(self): self.assertEqual(hitung_diskon(250000, True, "SALAH10"), 0)
    def test_22_voucher_kosong(self): self.assertEqual(hitung_diskon(250000, True, ""), 0)
    def test_23_negatif(self):
        with self.assertRaises(ValueError): hitung_diskon(-1, True, "POLINES10")

class TestCheckout(unittest.TestCase):
    def test_24_satu(self): self.assertTrue(validasi_checkout(["Laptop"]))
    def test_25_dua(self): self.assertTrue(validasi_checkout(["Laptop", "Mouse"]))
    def test_26_kosong(self): self.assertFalse(validasi_checkout([]))
    def test_27_none(self): self.assertFalse(validasi_checkout(None))
    def test_28_string(self): self.assertFalse(validasi_checkout("Laptop"))

loader = unittest.TestLoader()
suite = unittest.TestSuite()
for cls in (TestJumlah, TestSubtotal, TestDiskon, TestCheckout):
    suite.addTests(loader.loadTestsFromTestCase(cls))
result = unittest.TextTestRunner(verbosity=2).run(suite)
print("Total:", result.testsRun)
print("FAIL:", len(result.failures), "ERROR:", len(result.errors))
```

## SEL 3 — Demonstrasi defect (BUG-01)
Fungsi `validasi_jumlah_bug` memakai batas atas ≤ 11, sehingga test_11_harus_ditolak **FAIL**. FAIL ini disengaja sebagai bukti test dapat mendeteksi defect.

```python
import unittest

class TestDefect(unittest.TestCase):
    def test_11_harus_ditolak(self):
        self.assertFalse(validasi_jumlah_bug(11, 15))

suite_bug = unittest.TestLoader().loadTestsFromTestCase(TestDefect)
result_bug = unittest.TextTestRunner(verbosity=2).run(suite_bug)
print("FAIL yang diharapkan:", len(result_bug.failures))
```

## SEL 4 — Pembuktian perbaikan
Setelah batas diperbaiki menjadi ≤ 10 pada `validasi_jumlah`, test yang sama PASS. Jalankan kembali SEL 2 sebagai regression test penuh.

```python
import unittest

class TestPerbaikan(unittest.TestCase):
    def test_11_harus_ditolak(self):
        self.assertFalse(validasi_jumlah(11, 15))

suite_fix = unittest.TestLoader().loadTestsFromTestCase(TestPerbaikan)
result_fix = unittest.TextTestRunner(verbosity=2).run(suite_fix)
print("Perbaikan berhasil:", result_fix.wasSuccessful())
```

## SEL 5 — Enam test tambahan (TestTambahan)
Total pengujian normal: 28 + 6 = 34 test.

```python
import unittest

class TestTambahan(unittest.TestCase):
    def test_stok_tepat(self): self.assertTrue(validasi_jumlah(5, 5))
    def test_stok_kurang(self): self.assertFalse(validasi_jumlah(6, 5))
    def test_bool_ditolak(self): self.assertFalse(validasi_jumlah(True, 15))
    def test_diskon_batas(self):
        self.assertEqual(hitung_diskon(100000, True, "POLINES10"), 10000)
    def test_voucher_salah(self):
        self.assertEqual(hitung_diskon(250000, True, "SALAH10"), 0)
    def test_harga_negatif(self):
        with self.assertRaises(ValueError): hitung_subtotal(-1, 2)

suite_extra = unittest.TestLoader().loadTestsFromTestCase(TestTambahan)
result_extra = unittest.TextTestRunner(verbosity=2).run(suite_extra)
print("Tambahan:", result_extra.testsRun, "Berhasil:", result_extra.wasSuccessful())
```
